### Task 1: Arrays, Statistics and Indexing

```python
import numpy as np
import matplotlib.pyplot as plt

np.random.seed(42)
array_A = np.random.randint(1,101, size=100)
print(f"Array: {array_A}\n\n")

print(f"Min: {np.min(array_A)}")
print(f"Max: {np.max(array_A)}")
print(f"Median: {np.median(array_A)}")
print(f"Mean: {np.mean(array_A):.2f}")
print(f"Std Dev: {np.std(array_A):.2f}")

array_B = np.arange(100)

zeros_arr = np.zeros((3,3))
ones_arr = np.ones((2,4), dtype=np.float64)
print(f"\nZeros Array Shape: {zeros_arr.shape}, Data Type: {zeros_arr.dtype}")
print(zeros_arr)
print(f"Ones Array Shape: {ones_arr.shape}, Data Type: {ones_arr.dtype}")
print(ones_arr)

linespace_arr = np.linspace(0,10,num=5)
print(f"\nLinspace Array: {linespace_arr}")
```

### Task 2: Vectorized Arithmetic & Linear Algebra

```python
X = np.array([[2,3], [1,4]])
Y = np.array([[5,1], [0,2]])

print(f"Addition (X + Y):\n{X + Y}")
print(f"Element-wise Product (X * Y):\n{X * Y}")
print(f"Matrix Multiplication (X @ Y): \n{X @ Y}")

A = np.array([[3,1], [2,4]])
A_transpose = A.T
A_det = np.linalg.det(A)

print(f"\nMatrix A:\n{A}")
print(f"Transpose of A:\n{A_transpose}")
print(f"Determinant of A: {A_det:.2f}")

if A_det != 0:
    A_inv = np.linalg.inv(A)
    print(f"Inverse of A:\n{A_inv}")

    identity_check = A @ A_inv
    is_identity = np.allclose(identity_check, np.eye(2))
    print(f"A @ A^-1 Verification (is Identity? ): {is_identity}")
```

### Task 3: Normal Distribution and Histogram

```python
mu_chosen = 50
sigma_chosen = 8 
sample_size = 1000

norm_samples = np.random.normal(loc=mu_chosen, scale=sigma_chosen, size=sample_size)

sample_mean = np.mean(norm_samples)
sample_std = np.std(norm_samples)

print(f"Chosen Mean: {mu_chosen} | Sample Mean: {sample_mean:2f}")
print(f"Chosen Std: {sigma_chosen} | Sample Std: {sample_std:.2f}")

plt.figure(figsize=(8,5))
plt.hist(norm_samples, bins=30, density=True, alpha=0.6, color='g', edgecolor='black')

xmin, xmax = plt.xlim()
x = np.linspace(xmin, xmax, 100)
p = (1/ (sigma_chosen * np.sqrt(2*np.pi))) * np.exp(-0.5 * ((x - mu_chosen) / sigma_chosen) ** 2)

plt.plot(x,p, 'k', linewidth=2, label='Theoretical PDF')
plt.axvline(sample_mean, color='red', linestyle='dashed', linewidth=1.5, label=f'Sample Mean ({sample_mean:.2f})')
plt.title('Normal Distribution Hostogram (N=1000)')
plt.xlabel('Value')
plt.ylabel('Probability Density')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```
