### Task 2: Vectorized Arithmetic & Linear Algebra

```python
X = np.array([[2,3], [1,4]])
Y = np.array([[5,1], [0,2]])

print(f"Addition (X + Y):\n{X + Y}")
print(f"Element-wise Product (X * Y):\n{X * Y}")
print(f"Matrix Multiplication (X @ Y): \n{X @ Y}")

A = np.array([[3,1], [2,4]])
A_transpose = A.T
A_det = np.linalg.det(A)

print(f"\nMatrix A:\n{A}")
print(f"Transpose of A:\n{A_transpose}")
print(f"Determinant of A: {A_det:.2f}")

if A_det != 0:
    A_inv = np.linalg.inv(A)
    print(f"Inverse of A:\n{A_inv}")

    identity_check = A @ A_inv
    is_identity = np.allclose(identity_check, np.eye(2))
    print(f"A @ A^-1 Verification (is Identity? ): {is_identity}")
```

### Task 3: Normal Distribution and Histogram

```python
mu_chosen = 50
sigma_chosen = 8 
sample_size = 1000

norm_samples = np.random.normal(loc=mu_chosen, scale=sigma_chosen, size=sample_size)

sample_mean = np.mean(norm_samples)
sample_std = np.std(norm_samples)

print(f"Chosen Mean: {mu_chosen} | Sample Mean: {sample_mean:2f}")
print(f"Chosen Std: {sigma_chosen} | Sample Std: {sample_std:.2f}")

plt.figure(figsize=(8,5))
plt.hist(norm_samples, bins=30, density=True, alpha=0.6, color='g', edgecolor='black')

xmin, xmax = plt.xlim()
x = np.linspace(xmin, xmax, 100)
p = (1/ (sigma_chosen * np.sqrt(2*np.pi))) * np.exp(-0.5 * ((x - mu_chosen) / sigma_chosen) ** 2)

plt.plot(x,p, 'k', linewidth=2, label='Theoretical PDF')
plt.axvline(sample_mean, color='red', linestyle='dashed', linewidth=1.5, label=f'Sample Mean ({sample_mean:.2f})')
plt.title('Normal Distribution Hostogram (N=1000)')
plt.xlabel('Value')
plt.ylabel('Probability Density')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

## Part B: Data Wrangling with Pandas - Titanic Dataset

### Task 4: Load and Inspect Data

```python
import pandas as pd
df = pd.read_csv('train.csv')
print("1. First 5 Rows (head): ")
print(df.head())

print("\n2. Last 5 Rows (tail): ")
print(df.tail())

print(f"\n3. Shape (Rows, Columns): {df.shape}")

print("\n4. Column Names: ")
print(df.columns.tolist())

print("\n5. DataFrame info: ")
df.info()

print("\n6. Summary Statistics (describe): ")
print(df.describe())

print("\n loc vs iloc Selection Example")
loc_selection = df.loc[0:2, ['Name', 'Age', 'Pclass']]
print("Using loc (Rows with index labels 0 to 2, specific column names): \n", loc_selection)

iloc_selection = df.iloc[0:2, [3,5,2]]
print("\nUsing iloc (First 2 rows by position, columns at positions 3,5,2): \n", iloc_selection)
```

### Task 5: Filtering and Querying

```python
males_over_50 = df[(df['Sex'] == 'male') & (df['Age'] > 50)]
print(f"1. Male Passengers older than 50: {len(males_over_50)}")

female_pclass1 = df[(df['Sex'] == 'female') & (df['Pclass'] == 1)]
count_female_pclass1 = len(female_pclass1)
survival_rate_female_p1 = female_pclass1['Survived'].mean() * 100 if count_female_pclass1 > 0 else 0

print(f"2. Female 1st-class passengers: {count_female_pclass1}")
print(f"\tSurvival Percentage: {survival_rate_female_p1:.2f}")

overall_median_fare = df['Fare'].median()
q3_result = df.query("20 <= Age <= 40 and Fare > @overall_median_fare and Survived == 1")
print(f"3. Age 20-40, Fare > Overall median ({overall_median_fare:.2f}), and survived: {len(q3_result)}")

q4_result = df.query("SibSp == 0 and Parch == 0 and Age < 30 and Survived == 0")
print(f"4. Travelling alone, Age < 30, and did not survive: {len(q4_result)}")

s_median_fare = df[df['Embarked'] == 'S']['Fare'].median()
q5_result = df.query("Embarked == 'S' and Pclass in [2,3] and Fare > @s_median_fare")
print(f"5. Embarked 'S', Pclass 2/3, and Fare > Southampton median ({s_median_fare:.2f}): {len(q5_result)}")
```

### Task 6: groupby() and Aggregation

```python
survival_by_sex = df.groupby('Sex')['Survived'].mean()
print(f"1. Survival Rate by Sex: \n\t", survival_by_sex)

survival_by_pclass = df.groupby('Pclass')['Survived'].mean()
print("\n2. Survival Rate by Pclass: \n\t", survival_by_pclass)

avg_age_fare_pclass = df.groupby('Pclass')[['Age', 'Fare']].mean()
print(f"\n3. Average Age and Fare by Pclass: \n\t", avg_age_fare_pclass)

sex_pclass_agg = df.groupby(['Sex', 'Pclass'])['Survived'].agg(
    passenger_count='count',
    survival_rate='mean',
)
print("\n4. Passenger Count & Survival Rate by Sex-Pclass:\n\t", sex_pclass_agg)

embarked_agg = df.groupby('Embarked').agg(
    passenger_count=('PassengerId', 'count'),
    avg_fare=('Fare', 'mean'),
    survival_rate=('Survived', 'mean')
)
print("\n5. Summary by Embarked: \n\t", embarked_agg)
```

### Task 7: Missing Values and Fare Outliers

```python
missing_df = pd.DataFrame({
    'missing_count': df.isnull().sum(),
    'missing_percentage': (df.isnull().sum() / len(df)) * 100
})
print("Missing Value Summary:\n", missing_df)

plt.figure(figsize=(8, 4))
missing_df['missing_count'].plot(kind='bar', color='salmon', edgecolor='black')
plt.title('Missing Value Count per Column')
plt.xlabel('Columns')
plt.ylabel('Missing Values Count')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()

print(f"Age missing before mean imputation: {df['Age'].isnull().sum()}")
age_mean_imputed = df['Age'].fillna(df['Age'].mean())
print(f"Age missing after mean imputation: {age_mean_imputed.isnull().sum()}")

df_impute_demo = df.copy()

age_mean = df_impute_demo['Age'].fillna(df_impute_demo['Age'].mean())

age_median = df_impute_demo['Age'].fillna(df_impute_demo['Age'].median())

embarked_mode = df_impute_demo['Embarked'].fillna(df_impute_demo['Embarked'].mode()[0])

non_null_ages = df_impute_demo['Age'].dropna()
age_random = df_impute_demo['Age'].apply(
    lambda x: np.random.choice(non_null_ages) if pd.isnull(x) else x
)

print("\nImputation check:")
print(f"- Age (Mean Imputed) Missing: {age_mean.isnull().sum()}")
print(f"- Age (Median Imputed) Missing: {age_median.isnull().sum()}")
print(f"- Embarked (Mode Imputed) Missing: {embarked_mode.isnull().sum()}")
print(f"- Age (Random Imputed) Missing: {age_random.isnull().sum()}")

Q1 = df['Fare'].quantile(0.25)
Q3 = df['Fare'].quantile(0.75)
IQR = Q3 - Q1

lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

fare_outliers = df[(df['Fare'] < lower_bound) | (df['Fare'] > upper_bound)]

print("\nFare Outlier Analysis:")
print(f"Q1: {Q1:.2f} | Q3: {Q3:.2f} | IQR: {IQR:.2f}")
print(f"Lower Bound: {lower_bound:.2f} | Upper Bound: {upper_bound:.2f}")
print(f"Total Outliers in Fare: {len(fare_outliers)}")
```

### Task 8: Features and Pivot Table

```python
df['FamilySize'] = df['SibSp'] + df['Parch'] + 1
df['IsAlone'] = (df['FamilySize'] == 1).astype(int)

print("Engineered Features Sample:")
print(df[['SibSp', 'Parch', 'FamilySize', 'IsAlone']].head())

# 2. Pivot Table: rows=Sex, columns=Pclass, values=mean Survived
pivot_table = df.pivot_table(
    index='Sex',
    columns='Pclass',
    values='Survived',
    aggfunc='mean'
)
print("\nPivot Table (Survival Rate by Sex & Pclass):\n", pivot_table)

# Finding the highest and lowest survival groups
pivot_unstacked = pivot_table.unstack()
highest_group = pivot_unstacked.idxmax()
highest_val = pivot_unstacked.max()

lowest_group = pivot_unstacked.idxmin()
lowest_val = pivot_unstacked.min()

print(f"\nHighest Survival Group: Pclass {highest_group[0]}, {highest_group[1]} ({highest_val * 100:.2f}%)")
print(f"Lowest Survival Group:  Pclass {lowest_group[0]}, {lowest_group[1]} ({lowest_val * 100:.2f}%)")
```

### Task 9: Visualisation and Cleaned Data

```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

sns.set_theme(style="whitegrid")

df = pd.read_csv('train.csv')
df['FamilySize'] = df['SibSp'] + df['Parch'] + 1
df['IsAlone'] = (df['FamilySize'] == 1).astype(int)

os.makedirs('figures', exist_ok=True)

num_cols = ['Survived', 'Pclass', 'Age', 'SibSp', 'Parch', 'Fare', 'FamilySize', 'IsAlone']
corr_matrix = df[num_cols].corr()

plt.figure(figsize=(9, 7))
sns.heatmap(corr_matrix, annot=True, fmt=".2f", cmap="coolwarm", vmin=-1, vmax=1, linewidths=0.5)
plt.title('Correlation Heatmap of Numerical Features', fontsize=14, pad=12)
plt.tight_layout()
plt.savefig('figures/correlation_heatmap.png', dpi=300)
plt.show()

corr_pairs = corr_matrix.unstack().sort_values()
corr_pairs = corr_pairs[corr_pairs < 1.0]

print(f"Strongest Negative Correlation: {corr_pairs.index[0]} -> {corr_pairs.iloc[0]:.3f}")
print(f"Strongest Positive Correlation: {corr_pairs.index[-1]} -> {corr_pairs.iloc[-1]:.3f}")


survival_by_sex = df.groupby('Sex')['Survived'].mean() * 100

plt.figure(figsize=(6, 5))
ax = sns.barplot(x=survival_by_sex.index, y=survival_by_sex.values, palette=['#e74c3c', '#3498db'])
plt.title('Survival Rate by Gender', fontsize=14)
plt.xlabel('Sex', fontsize=12)
plt.ylabel('Survival Rate (%)', fontsize=12)
plt.ylim(0, 100)

for p in ax.patches:
    ax.annotate(f"{p.get_height():.2f}%", 
                (p.get_x() + p.get_width() / 2., p.get_height() / 2),
                ha='center', va='center', color='white', fontweight='bold', fontsize=12)

plt.tight_layout()
plt.savefig('figures/survival_by_sex.png', dpi=300)
plt.show()

print(f"Female Survival Rate: {survival_by_sex['female']:.2f}%")
print(f"Male Survival Rate:   {survival_by_sex['male']:.2f}%")

plt.figure(figsize=(10, 6))
sns.scatterplot(
    data=df, 
    x='Age', 
    y='Fare', 
    hue='Survived', 
    palette={0: '#e74c3c', 1: '#2ecc71'},
    alpha=0.7, 
    s=60
)
plt.title('Age vs. Fare (Distinguished by Survival)', fontsize=14)
plt.xlabel('Age', fontsize=12)
plt.ylabel('Fare ($)', fontsize=12)
plt.legend(title='Status', labels=['Did not survive (0)', 'Survived (1)'])
plt.tight_layout()
plt.savefig('figures/age_vs_fare.png', dpi=300)
plt.show()

df_cleaned = df.copy()
df_cleaned['Age'] = df_cleaned['Age'].fillna(df_cleaned['Age'].median())
df_cleaned['Embarked'] = df_cleaned['Embarked'].fillna(df_cleaned['Embarked'].mode()[0])
df_cleaned.drop(columns=['Cabin'], inplace=True) # Cabin has >77% missing values

df_cleaned.to_csv('titanic_cleaned.csv', index=False)
print("\nCleaned dataset exported to 'titanic_cleaned.csv'")
```
